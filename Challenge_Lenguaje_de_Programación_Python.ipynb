Dentro de la siguiente dinámica, **se creará una agenda de cinco contactos** haciendo uso de las **estructuras de datos** utilizando **Python**.

Donde los datos que debe contener cada contacto de la angenda son:

*   **Nombre**
*   **Teléfono**
*   **Correo**

```python
def createContacts(_numberOfContacts: int):
    for i in range(_numberOfContacts):
        name = input("Ingrese nombre de contacto:")
        phone = input("Ingrese numero de telefono:")
        email = input("Ingrese el correo electronico:")
        contacto = {}
        contacto["nombre"] = name
        contacto["telefono"] = phone
        contacto["correo"] = email
        agenda.append(contacto)

def printContacts():
    print("##### Contactos ######")
    for contacto in agenda:
        print("Nombre:", contacto['nombre'])
        print("Telefono:", contacto['telefono'])
        print("Correo:", contacto['correo'])
        print("------------------------------------")


### main ###

agenda = []
createContacts(5)
printContacts()
```
