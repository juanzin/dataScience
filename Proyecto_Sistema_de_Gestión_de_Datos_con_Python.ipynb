```python
''' 
DESCRIPCION:
el objetivo de este programa es administrar los productos de una tienda, cada producto tendra las siguientes propiedades:
    * nombre
    * codigo
    * precio
    * cantidad (numero de piezas disponibles)

    operaciones:
    * agregar producto
    * eliminar producto
    * ver lista de productos
    * actualizar productos
    * cerrar menu
'''
```

```python
# product definition
class Product:
    def __init__(self, name, price, code, quantity):
        self.name = name
        self.price = price
        self.code = code
        self.quantity = quantity
    
    def set_name(self, new_name: str):
        self.name = new_name

    def set_code(self, new_code: int):
        self.code = new_code

    def set_price(self, new_price: int):
        self.price = new_price

    def set_quantity(self, new_quantity: int):
        self.quantity = new_quantity
    
    def print_product(self):
        print("--------------------------")
        print("name:", self.name)
        print("code:", self.code)
        print("price:", self.price)
        print("quantity:", self.quantity)
        print("--------------------------")




        
```

```python
class Catalog:
    def __init__(self):
        self.product_list = []

    def add_product(self):
        name = input("Enter product name")
        price = input("Enter product price")
        code = input("Enter product code")
        quantity = input("Enter product quantity")
        product = Product(name, price, code, quantity)
        self.product_list.append(product)

    def remove_product(self):
        name = input("Enter product name")
        index = self._get_index_product_by_product_name(name)
        if (index != -1):
            self.product_list.pop(index)
            print("Product removed succesfully!")

    def update_product(self):
        name = input("Enter product name")
        index = self._get_index_product_by_product_name(name)
        if (index != -1):
            product = self.product_list[index]
            self._display_menu_to_update_product()
            option = input("Select an option to update")
            while(option != 5):
                match option:
                    case '1':
                        name = input("Enter the new product name")
                        product.set_name(name)
                    case '2':
                        price = input("Enter the new product price")
                        product.set_price(price)
                    case '3':
                        code = input("Enter the new product code")
                        product.set_code(code)
                    case '4':
                        quantity = input("Enter the new product quantity")
                        product.set_quantity(quantity)
                    case '5':
                        break
                    case _:
                        print("Unknown option")
                        break
                self._display_menu_to_update_product()
                option = input("Select an option to update")

    
    def print_catalog(self):
        if len(self.product_list) == 0:
            print("no products available")
            return

        print("###     PRODUCTS    ###")
        for product in self.product_list:
            product.print_product()

    def _display_menu_to_update_product(self):
        print("###################################")
        print("## What do you want to update?   ##")
        print("## 1. Update product name        ##")
        print("## 2. Update product price       ##")
        print("## 3. Update product code        ##")
        print("## 4. Update product quantity    ##")
        print("## 5. Close menu                 ##")
        print("###################################")

    def _get_index_product_by_product_name(self, name: str) -> int:
        index: int = -1
        itemFound: bool = False
        for product in self.product_list:
            index += 1
            if name.lower() == product.name.lower():
                itemFound = True
                break
        if itemFound == False:
            index = -1
            print("product not found")

        return index
```

```python
## product manager
def display_menu():
    print("########################")
    print("## Select an option   ##")
    print("## 1. Display catalog ##")
    print("## 2. Add product     ##")
    print("## 3. Delete product  ##")
    print("## 4. Update product  ##")
    print("## 5. Close Menu      ##")
    print("########################")

my_catalog = Catalog()
display_menu()
option = input("Select an option")

while(option != 5):
    match option:
        case '1':
            my_catalog.print_catalog()
        case '2':
            my_catalog.add_product()
        case '3':
            my_catalog.remove_product()
        case '4':
            my_catalog.update_product()
        case '5':
            break
        case _:
            print("Unknown option")
            break

    display_menu()
    option = input("Select an option")
```
